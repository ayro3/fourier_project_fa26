```python
import yfinance as yf
import matplotlib.pyplot as plt
```

```python
appl = yf.Ticker("AAPL") # intialize ticker object
df_appl_his = appl.history(period = "max") # get full asset history
close_price = df_appl_his.Close # close prices and dates as a series obj
```

```python
plt.plot(close_price.index,close_price) #plotting price over time

plt.title("Apple Asset Price from 1980 to 2026")
plt.xlabel("Date")
plt.ylabel("Price")
plt.show()
```
